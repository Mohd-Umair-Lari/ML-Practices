```python
import os
from pathlib import Path
from dotenv import load_dotenv
```

```python
env_path='../.env'
load_dotenv(dotenv_path=env_path)
key=os.getenv("AI_STUDIO_KEY")
```

```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled

from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)

from langchain_community.vectorstores import FAISS

from langchain_core.prompts import PromptTemplate
```

```python
# STEP 1(a)--INDEXING(DOCUMNET INGESTION)

video_id="LVzBMUwUkK8"
try:
    # if u don't care which language, this returns the "Best" One
    ytt_api=YouTubeTranscriptApi()
    transcript_list=ytt_api.fetch(
        video_id,
        languages=["en"]
    )

    # flatten it to plain text
    transcript=''.join(chunk.text for chunk in transcript_list)
    print(transcript)

except TranscriptsDisabled:
    print("No captions available for this video.")
```

```python
transcript_list
```

```python
## Stet 1(b)---INDEXING(tEXT SPLITTING)
splitter=RecursiveCharacterTextSplitter(chunk_size=1000,chunk_overlap=200)
chunks=splitter.create_documents([transcript])
```

```python
len(chunks)
```

```python
chunks[0]
```

```python
## STEP1C & 1D -- INDEXING(EMBEDDING GENERATION AND STORING IN VECTOR STORE)
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import FAISS

embedding = GoogleGenerativeAIEmbeddings(
    model="gemini-embedding-001",
    api_key=key
)

vector_store = FAISS.from_documents(
    chunks,
    embedding
)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(["057d63d3-2a96-4da2-9c4b-1b9b746855ae"])
```

```python
### STEP 2--> RETRIEVAL
retriever=vector_store.as_retriever(search_type="similarity", search_kwargs={"k":4})
```

```python
retriever.invoke("what is Generative Ai")
```

```python
### STEP- 3 ---> AUGMENTATION
llm=ChatGoogleGenerativeAI(model="gemini-3.5-flash",temperature=0.2,api_key=key)
```

```python
prompt=PromptTemplate(
    template='''ArithmeticErroryou are a helpful assistant.
    Answer ONLY from the provided tanscript context.If the context is insufficient, justify say you don,t know.

    {context}
    Question:{question}
    ''',
    input_variable=['context','question']
)
```

```python
question="What is Pattern Recognition"
retrieved_docs=retriever.invoke(question)
```

```python
retrieved_docs
```

```python
context_text="\n".join(doc.page_content for doc in retrieved_docs)
print(context_text)
```

```python
final_prompt=prompt.invoke({"context":context_text,"question":question})
final_prompt.type
```

```python
print(final_prompt.to_string())
```

```python
### STEP-4 --> GENERATION
answer=llm.invoke(final_prompt)
print(answer.content)
```

```python
## BUILDING A CHAIN
from langchain_core.runnables import RunnableParallel,RunnablePassthrough,RunnableLambda
from langchain_core.output_parsers import StrOutputParser
```

```python
parallel_chain=RunnableParallel({
    'context': lambda x: context_text,
    'question': RunnablePassthrough()
})
```

```python
print(parallel_chain.invoke(" what is Pattern recognition ?"))
```

```python
parser=StrOutputParser()
```

```python
main_chain=parallel_chain | prompt | llm | parser
```

```python
main_chain.invoke("What is Pattern Recognition?")
```
