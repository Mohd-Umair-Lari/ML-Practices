```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import Birch, KMeans
from sklearn.metrics import silhouette_score, davies_bouldin_score

df = pd.read_csv("Wholesale_customers_data.csv")
X = df.select_dtypes(include=['int64','float64'])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

birch_model = Birch(branching_factor=50, threshold=1.5, n_clusters=None)
birch_model.fit(X_scaled)
labels_birch = birch_model.labels_

leaf_centroids = birch_model.subcluster_centers_

kmeans_model = KMeans(n_clusters=len(leaf_centroids),
                      init=leaf_centroids,
                      n_init=1,
                      random_state=42)
kmeans_model.fit(X_scaled)
labels_birch_kmeans = kmeans_model.labels_

sil_birch = silhouette_score(X_scaled, labels_birch)
dbi_birch = davies_bouldin_score(X_scaled, labels_birch)

sil_birch_kmeans = silhouette_score(X_scaled, labels_birch_kmeans)
dbi_birch_kmeans = davies_bouldin_score(X_scaled, labels_birch_kmeans)

print("=== Clustering Performance ===")
print("Pure BIRCH:")
print(f"  Silhouette Score: {sil_birch:.4f}")
print(f"  Davies-Bouldin Index: {dbi_birch:.4f}")

print("\nBIRCH + KMeans refinement:")
print(f"  Silhouette Score: {sil_birch_kmeans:.4f}")
print(f"  Davies-Bouldin Index: {dbi_birch_kmeans:.4f}")
```
