```python
import pandas as pd
import numpy as np

df=pd.read_csv("1.csv")
```

```python
train_df = df[df['EXAM SCORE'].notna()]

x = train_df['STUDY HOURS'].values
y = train_df['EXAM SCORE'].values
n = len(x)

sum_x = np.sum(x)
sum_y = np.sum(y)
sum_xy = np.sum(x * y)
sum_x2 = np.sum(x * x)

m = (n * sum_xy - sum_x * sum_y) / (n * sum_x2 - sum_x**2)
c = (sum_y - m * sum_x) / n

print(f"Linear Regression Equation: Exam Score = {m:.2f} * Study Hours + {c:.2f}")

for i in range(len(df)):
    if pd.isna(df.loc[i, 'EXAM SCORE']):
        df.loc[i, 'EXAM SCORE'] = m * df.loc[i, 'STUDY HOURS'] + c

print("\nPredicted using Linear Regression:\n", df)
```
