```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from PIL import Image

image = Image.open('DOGCAT.webp')

train_data = pd.DataFrame({
    'Ear Shape': [0.2, 0.25, 0.85, 0.9],
    'Fluffiness': [0.3, 0.35, 0.8, 0.9],
    'Label': [0, 0, 1, 1]
})

X_train = train_data[['Ear Shape', 'Fluffiness']]
y_train = train_data['Label']

model = LogisticRegression()
model.fit(X_train, y_train)

X_test = pd.DataFrame({
    'Ear Shape': [0.2, 0.85],
    'Fluffiness': [0.3, 0.8]
})

y_actual = [0, 1]
y_pred = model.predict(X_test)

results = pd.DataFrame({
    'Actual': ['Cat' if x == 0 else 'Dog' for x in y_actual],
    'Predicted': ['Cat' if x == 0 else 'Dog' for x in y_pred]
})

print(results)
```
