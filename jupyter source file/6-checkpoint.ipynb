```python
import numpy as np

def kmeans(X, k, max_iters=100):
    np.random.seed(42)
    centroids = X[np.random.choice(len(X), k, replace=False)]
    
    for iteration in range(max_iters):
        distances = np.array([[np.linalg.norm(x - c) for c in centroids] for x in X])
        clusters = np.argmin(distances, axis=1)
        sse = 0
        for i in range(k):
            cluster_points = X[clusters == i]
            if len(cluster_points) > 0:
                sse += np.sum((cluster_points - centroids[i])**2)
        print(f"Iteration {iteration + 1}, SSE: {sse}")
        new_centroids = np.array([X[clusters == i].mean(axis=0) if len(X[clusters == i]) > 0 else centroids[i] for i in range(k)])        
        if np.all(centroids == new_centroids):
            break
        centroids = new_centroids
    return clusters, centroids, sse

X = np.array([[1, 2], [1, 4], [1, 0],
              [4, 2], [4, 4], [4, 0]])
clusters, centroids, sse = kmeans(X, k=2)
```

```python
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

X = np.array([[1, 2], [1, 4], [1, 0],
              [4, 2], [4, 4], [4, 0]])

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

k = 3
max_iters = 10
kmeans = KMeans(n_clusters=k, max_iter=1, init='random', n_init=1, random_state=42, verbose=0)

sse_list = []
clusters_list = []

for i in range(max_iters):
    kmeans.max_iter = 1
    kmeans.fit(X_scaled)
    
    clusters = kmeans.labels_
    centers = kmeans.cluster_centers_
    sse = 0
    for cluster_id in range(k):
        cluster_points = X_scaled[clusters == cluster_id]
        sse += np.sum((cluster_points - centers[cluster_id])**2)
    
    sse_list.append(sse)
    clusters_list.append(clusters)
    
    print(f"Iteration {i+1}, SSE: {sse}, Clusters: {clusters}")
    kmeans.init = centers

print("\nFinal SSE values per iteration:", sse_list)
```
