```python
import cv2
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

def k_means_image_segmentation(image_path, n_clusters):
    image = cv2.imread("images.jpg")
    if image is None:
        print(f"Error: Could not load image from {image}")
        return None
    image_rgb = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    pixels = image_rgb.reshape(-1, 3)

    pixels = np.float32(pixels)

    kmeans = KMeans(n_clusters=n_clusters, n_init='auto', random_state=42)
    kmeans.fit(pixels)
    centers = np.uint8(kmeans.cluster_centers_)
    labels = kmeans.labels_

    segmented_image = centers[labels].reshape(image_rgb.shape)

    return segmented_image


input_image_path = 'images.jpg' 
num_clusters = 2 
segmented_img = k_means_image_segmentation(input_image_path, num_clusters)

if segmented_img is not None:
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.imshow(cv2.imread(input_image_path))
    plt.title('Original Image')
    plt.axis('off')

    plt.subplot(1, 2, 2)
    plt.imshow(segmented_img)
    plt.title(f'Segmented Image (K={num_clusters})')
    plt.axis('off')

    plt.show()
```
