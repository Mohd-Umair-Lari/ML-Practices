```python
# Apriori Algorithm

import pandas as pd
from itertools import combinations
from collections import defaultdict

df = pd.read_csv("data.csv")
transactions = df.apply(lambda row: set(row.dropna()), axis=1).tolist()
total_transactions = len(transactions)
min_support = 0.2 * total_transactions
min_confidence = 0.3
```

```python
def find_frequent_items(transactions, min_support):
    item_counts = defaultdict(int)
    for basket in transactions:
        for item in basket:
            item_counts[frozenset([item])] += 1
    frequent_items = {item: count for item, count in item_counts.items() if count >= min_support}
    all_frequent = dict(frequent_items)
    size = 2
    while frequent_items:
        new_counts = defaultdict(int)
        items_list = list(frequent_items.keys())
        for i in range(len(items_list)):
            for j in range(i + 1, len(items_list)):
                new_set = items_list[i] | items_list[j]
                if len(new_set) == size:
                    for basket in transactions:
                        if new_set.issubset(basket):
                            new_counts[new_set] += 1
        
        frequent_items = {item: count for item, count in new_counts.items() if count >= min_support}
        all_frequent.update(frequent_items)
        size += 1
    return all_frequent

def make_rules(frequent_itemsets, min_confidence):
    rules = []
    support_map = {item: count for item, count in frequent_itemsets.items()}
    for itemset in frequent_itemsets:
        if len(itemset) < 2:
            continue
        for i in range(1, len(itemset)):
            for part in combinations(itemset, i):
                antecedent = frozenset(part)
                consequent = itemset - antecedent
                if antecedent in support_map:
                    confidence = support_map[itemset] / support_map[antecedent]
                    if confidence >= min_confidence:
                        rules.append((list(antecedent), list(consequent), round(confidence, 2)))
    return rules

frequent_sets = find_frequent_items(transactions, min_support)
rules = make_rules(frequent_sets, min_confidence)

for lhs, rhs, confidence in rules[:10]:
    print(f"{lhs} => {rhs} (Confidence: {confidence})")
```

```python
# FP Gowth Algorithm:

import pandas as pd
from collections import defaultdict

df = pd.read_csv("data.csv")
transactions = df.apply(lambda row: list(row.dropna()), axis=1).tolist()
min_support_count = int(0.2 * len(transactions))

class FPTreeNode:
    def __init__(self, item, count, parent):
        self.item = item
        self.count = count
        self.parent = parent
        self.children = {}
        self.link = None
```

```python
def build_fptree(transactions, min_support):
    item_counts = defaultdict(int)
    for transaction in transactions:
        for item in transaction:
            item_counts[item] += 1
    item_counts = {item: count for item, count in item_counts.items() if count >= min_support}
    if not item_counts:
        return None, None
    header_table = {item: [count, None] for item, count in item_counts.items()}
    root = FPTreeNode(None, 1, None)
    for transaction in transactions:
        items = [item for item in transaction if item in item_counts]
        items.sort(key=lambda item: item_counts[item], reverse=True)
        current = root
        for item in items:
            if item in current.children:
                current.children[item].count += 1
            else:
                new_node = FPTreeNode(item, 1, current)
                current.children[item] = new_node
                if header_table[item][1] is None:
                    header_table[item][1] = new_node
                else:
                    link_node = header_table[item][1]
                    while link_node.link:
                        link_node = link_node.link
                    link_node.link = new_node
            current = current.children[item]
    return root, header_table

def ascend_fp_tree(node):
    path = []
    while node and node.parent and node.parent.item is not None:
        node = node.parent
        path.append(node.item)
    return path

def find_prefix_paths(base_pat, node):
    paths = []
    while node:
        path = ascend_fp_tree(node)
        if path:
            paths.append((path, node.count))
        node = node.link
    return paths

def mine_fptree(header_table, min_support, prefix, frequent_patterns):
    sorted_items = sorted(header_table.items(), key=lambda x: x[1][0])
    for base_pat, (count, node) in sorted_items:
        new_freq_set = prefix.copy()
        new_freq_set.add(base_pat)
        frequent_patterns[frozenset(new_freq_set)] = count
        conditional_patterns = find_prefix_paths(base_pat, node)
        cond_trans = []
        for path, count in conditional_patterns:
            cond_trans.extend([path] * count)
        cond_tree, cond_header = build_fptree(cond_trans, min_support)
        if cond_header:
            mine_fptree(cond_header, min_support, new_freq_set, frequent_patterns)

def generate_rules(frequent_patterns, min_confidence, transaction_count):
    rules = []
    support_map = {itemset: count for itemset, count in frequent_patterns.items()}
    for itemset in frequent_patterns:
        if len(itemset) < 2:
            continue
        for i in range(1, len(itemset)):
            for antecedent in combinations(itemset, i):
                antecedent = frozenset(antecedent)
                consequent = itemset - antecedent
                if consequent and antecedent in support_map:
                    conf = support_map[itemset] / support_map[antecedent]
                    if conf >= min_confidence:
                        rules.append((list(antecedent), list(consequent), round(conf, 2)))
    return rules

frequent_patterns = {}
tree, header = build_fptree(transactions, min_support_count)
if header:
    mine_fptree(header, min_support_count, set(), frequent_patterns)

rules = generate_rules(frequent_patterns, min_confidence, len(transactions))

for X, Y, conf in rules[:10]:
    print(f"{X} -> {Y} | Confidence: {conf}")
```
