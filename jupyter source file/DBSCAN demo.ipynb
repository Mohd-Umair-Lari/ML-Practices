```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
```

```python
df=pd.read_csv("student_clustering.csv")
```

```python
df
```

```python
db=DBSCAN(eps=2, min_samples=5)
y=db.fit_predict(df)
y
```

```python
label=db.labels_
```

```python
plt.scatter(df.iloc[:,0],df.iloc[:,1],c=label)
```
