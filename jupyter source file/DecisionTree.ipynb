```python
from sklearn.datasets import load_iris
import dtreeviz
import graphviz.backend as be
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import confusion_matrix, accuracy_score
```

```python
clas=DecisionTreeClassifier()
iris=load_iris()
X=load_iris().data
y=load_iris().target
```

```python
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=.25,random_state=32)
```

```python
clas.fit(X_train,y_train)
y_pred=clas.predict(X_test)
print("The Confusion matrix is as:\n",confusion_matrix(y_test,y_pred))
print("The Accuracy score is as:\n",accuracy_score(y_test,y_pred))
```

```python
plot_tree(clas)
```

```python
viz=dtreeviz.model(
    clas,
    X_train,
    y_train,
    feature_names=iris.feature_names,
    class_names=["setosa", "versicolor", "virginica"]
)

viz.view()
```
