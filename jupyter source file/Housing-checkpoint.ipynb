**Gradien Desent with iterations**

```python
def gradient_descent(X, y, learning_rate=0.01, iterations=100):
    m, n=X.shape
    theta=np.zeros((n, 1))
    for i in range(iterations):
        prediction=X@theta
        
        loss=(prediction-y)
        gradient=(1/m)*(X.T@loss)
        
        theta=theta-(learning_rate*gradient)
    return theta

theta_gd = gradient_descent(X_test, y_test)
print("Gradient Descent Parameters:\n",theta_gd)
```

**Normal Equation**

```python
def normeqn(X,y):
    X_transpose = X.T
    X_transpose_X = np.dot(X_transpose, X)
    X_transpose_y = np.dot(X_transpose, y)
    X_with_intercept = X_transpose_X + np.eye(X_transpose_X.shape[0])
    
    theta = np.linalg.solve(X_with_intercept, X_transpose_y)
    
    return theta

theta_ne=normeqn(X_test,y_test)
print("Normal Equation Parameters:\n",theta_ne)
```
