```python
# BFS from given source s
from collections import deque

def bfsOfGraph(adj, s, visited, res):
    
    # Create a queue for BFS
    q = deque()
    
    # Mark source node as visited and enqueue it
    visited[s] = True
    q.append(s)
    
    # Iterate over the queue
    while q:
        
        # Dequeue a vertex and store it
        curr = q.popleft()
        res.append(curr)
        
        # Get all adjacent vertices of the dequeued 
        # vertex curr If an adjacent has not been 
        # visited, mark it visited and enqueue it
        for x in adj[curr]:
            if not visited[x]:
                visited[x] = True
                q.append(x)
    return res

# Perform BFS for the entire graph which maybe
# disconnected
def bfsDisconnected(adj):
    V = len(adj)
    
    # create an array to store the traversal
    res = []
    
    # Initially mark all the vertices as not visited
    visited = [False] * V
    
    # perform BFS for each node
    for i in range(V):
        if not visited[i]:
            bfsOfGraph(adj, i, visited, res)
    return res

if __name__ == "__main__":
    adj = [[1, 2], [0], [0],
        [4], [3, 5], [4]]
    ans = bfsDisconnected(adj)
    for i in ans:
        print(i, end=" ")
```

```python
def TSP(dist):
    n = len(dist)
    dp = [[float('inf')] * n for _ in range(1 << n)]
    parent = [[-1] * n for _ in range(1 << n)]
    dp[1][0] = 0
    for mask in range(1, 1 << n):
        for i in range(n):
            if mask & (1 << i):
                for j in range(n):
                    if not (mask & (1 << j)):
                        new_mask = mask | (1 << j)
                        if dp[new_mask][j] > dp[mask][i] + dist[i][j]:
                            dp[new_mask][j] = dp[mask][i] + dist[i][j]
                            parent[new_mask][j] = i
    result = float('inf')
    last_city = -1
    for i in range(1, n):
        if result > dp[(1 << n) - 1][i] + dist[i][0]:
            result = dp[(1 << n) - 1][i] + dist[i][0]
            last_city = i
    path = [0]
    mask = (1 << n) - 1
    while last_city != -1:
        path.append(last_city)
        new_mask = mask & ~(1 << last_city)
        last_city = parent[mask][last_city]
        mask = new_mask
    return result, path
dist = [
    [0, 10, 15, 20],
    [10, 0, 35, 25],
    [15, 35, 0, 30],
    [20, 25, 30, 0]
]
cost, path = TSP(dist)
print("Minimum cost to complete TSP:", cost)
print("Path:", " -> ".join(str(city) for city in path))
```
