```python
X = [2, 3, 4, 6, 7, 8, 10, 11]
Y = [50, 55, 58, 65, 70, 72, 80, 85]
mean_x = sum(X) / len(X)
mean_y = sum(Y) / len(Y)
numerator = sum((x - mean_x) * (y - mean_y) for x, y in zip(X, Y))
denominator = sum((x - mean_x) ** 2 for x in X)
m = numerator / denominator
c = mean_y - m * mean_x
def predict(x):
    return round(m * x + c)
missing_at_5 = predict(5)
missing_at_9 = predict(9)
print("Linear Regression Predictions:")
print(f"At Study Hour 5 → Predicted Score: {missing_at_5}")
print(f"At Study Hour 9 → Predicted Score: {missing_at_9}")
```

```python
def mean_prediction(x, x_list, y_list):
    for i in range(len(x_list) - 1):
        if x_list[i] < x < x_list[i + 1]:
            y1, y2 = y_list[i], y_list[i + 1]
            return (y1 + y2) / 2
    return None
x_list = [2, 3, 4, 6, 7, 8, 10, 11]
y_list = [50, 55, 58, 65, 70, 72, 80, 85]
pred_5 = mean_prediction(5, x_list, y_list)
pred_9 = mean_prediction(9, x_list, y_list)
print("\nMean-Based Predictions:")
print(f"At Study Hour 5 → Predicted Score: {pred_5}")
print(f"At Study Hour 9 → Predicted Score: {pred_9}")
```

```python
import csv
filename = '1.csv'
X, Y = [], []
missing_indices = []

with open(filename, 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        x = int(row['STUDY HOURS'])
        y = row['EXAM SCORE']
        if y.strip() == '':
            missing_indices.append(x)
        else:
            X.append(x)
            Y.append(float(y))
mean_x = sum(X) / len(X)
mean_y = sum(Y) / len(Y)

numerator = sum((x - mean_x) * (y - mean_y) for x, y in zip(X, Y))
denominator = sum((x - mean_x) ** 2 for x in X)
m = numerator / denominator
c = mean_y - m * mean_x
def predict_lr(x):
    return round(m * x + c, 2)
def mean_neighbor_estimate(x, x_known, y_known):
    for i in range(len(x_known) - 1):
        if x_known[i] < x < x_known[i + 1]:
            return round((y_known[i] + y_known[i + 1]) / 2, 2)
    return None
print("Missing value predictions:\n")

for x in missing_indices:
    lr_pred = predict_lr(x)
    mean_pred = mean_neighbor_estimate(x, X, Y)
    print(f"Study Hour: {x}")
    print(f" - Linear Regression: {lr_pred}")
    print(f" - Mean of Neighbors: {mean_pred}\n")
```
