```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
from itertools import combinations

# Sample dataset
dataset = [
    ['milk', 'bread', 'nuts', 'apple'],
    ['milk', 'bread', 'cornflakes'],
    ['milk', 'bread', 'nuts'],
    ['bread', 'cornflakes', 'apple'],
    ['milk', 'bread', 'cornflakes', 'apple']
]

# One-hot encoding
te = TransactionEncoder()
df = pd.DataFrame(te.fit(dataset).transform(dataset), columns=te.columns_)

# Apriori Algorithm
frequent_itemsets = apriori(df, min_support=0.4, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)
print("Apriori Itemsets:\n", frequent_itemsets)
print("\nAssociation Rules:\n", rules[['antecedents','consequents','support','confidence','lift']])

# Naive Algorithm
def naive(df, min_support=0.4):
    res, n = [], len(df)
    for r in range(1, len(df.columns)+1):
        for c in combinations(df.columns, r):
            sup = df[list(c)].all(axis=1).sum()/n
            if sup >= min_support:
                res.append({"itemsets": c, "support": sup})
    return pd.DataFrame(res)

print("\nNaive Itemsets:\n", naive(df))
```

```python
import pandas as pd
data = {
    "Antecedent": ["A","A","A","A","B","B","B"],
    "Consequent": [0,0,1,0,1,0,1]
}
df = pd.DataFrame(data)
total = len(df)
df,total
```

```python
lhs_count = len(df[df["Antecedent"]=="A"])
both_count = len(df[(df["Antecedent"]=="A") & (df["Consequent"]==0)])
print(both_count)
rhs_count = len(df[df["Consequent"]==0])
print(rhs_count)
support_A0 = both_count / total
print(support_A0)
confidence_A0 = both_count / lhs_count
lift_A0 = confidence_A0 / (rhs_count / total)
print("A → 0:", support_A0, confidence_A0, lift_A0)
```
