```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.impute import KNNImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.read_csv("Bengaluru_House_Data.csv")
df
df.shape
```

```python
df.isnull().sum()
```

```python
df['size']=df['size'].str.extract('(\d+\.?\d*)')
```

```python
df
```

```python
imputer=SimpleImputer(strategy='most_frequent')
df["size"]=imputer.fit_transform(df[["size"]]).ravel()
df["bath"]=imputer.fit_transform(df[["bath"]]).ravel()
df["balcony"]=imputer.fit_transform(df[["balcony"]]).ravel()
df["society"]=imputer.fit_transform(df[["society"]]).ravel()
```
