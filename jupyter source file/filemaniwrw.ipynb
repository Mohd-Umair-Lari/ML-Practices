```python
data=[
    ['Name', 'Roll', 'Math','Chem','Phy','Eng'],
    ['a',355,85,65,55,63],
    ['b',356,78,71,85,45],
    ['c',3,98,55,56,65],
    ['d',4,84,96,74,52],
    ['e',5,89,96,74,63],
    ['f',6,63,23,45,15]
]
with open("student.csv","r+")as obj:
    list=csv.writer(obj)
    list.writerows(data)
```

```python
with open("student.csv", mode="r") as file:
    reader= csv.DictReader(file)
    da=[row for row in reader]
for student in da:
    print(student)
```

```python
m=[]
c=[]
p=[]
e=[]
for student in da:
        m.append(int(student['Math'])),
        c.append(int(student['Chem'])),
        p.append(int(student['Phy'])),
        e.append(int(student['Eng']))
print(f"Math -> high {max(m)} , low {min(m)}")
print(f"CHEM -> high {max(c)} , low {min(c)}")
print(f"PHY -> high {max(p)} , low {min(p)}")
print(f"ENG -> high {max(e)} , low {min(e)}")
```
