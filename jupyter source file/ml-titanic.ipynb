```python
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
```

```python
train_data=pd.read_csv("titanic_train.csv")
train_data=train_data.dropna()
train_data.head(5)
```

```python
test_data=pd.read_csv("titanic_test.csv")
test_data=test_data.dropna()
test_data.head(5)
```

```python
women = train_data.loc[train_data.Sex == 'female']["Survived"]
rate_women = sum(women)/len(women)
print("% of women who survived:", rate_women)
```

```python
men=train_data.loc[train_data.Sex=='male']["Survived"]
rate_men=sum(men)/len(men)
print("%  oof men wh survived:", rate_men)
```

```python
y=train_data["Survived"]
features=["Pclass","Sex","Age","Parch"]
x=pd.get_dummies(train_data[features]) # substitutes the string values with numeric values
x_test=pd.get_dummies(test_data[features]) # substitutes the strings with numeric values
```

```python
clf=RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
clf.fit(x,y)
predictions=clf.predict(x_test)
```

```python
output = pd.DataFrame({'PassengerId': test_data.PassengerId, 'Survived': predictions})
output.to_csv('submission.csv', index=False)
print("Your submission was successfully saved!")
```
