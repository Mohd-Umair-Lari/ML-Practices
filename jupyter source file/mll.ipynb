```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
def load_data(filename):
    df = pd.read_csv(filename)
    return df
```

```python
def apply_kmeans(data, k):
    kmeans = KMeans(n_clusters=k, max_iter=300, random_state=42)
    kmeans.fit(data)
    centroids = kmeans.cluster_centers_
    labels = kmeans.labels_
    sse = kmeans.inertia_
    return kmeans, centroids, labels, sse
```

```python
filename = 'cluster_data.csv'
data = load_data(filename)
data_values = data.values
```

```python
k = 3
kmeans_model, final_centroids, cluster_labels, sse = apply_kmeans(data_values, k)
```

```python
print("Final Centroids:")
print(final_centroids)
```

```python
print("\nCluster Labels (Assignments for each point):")
print(cluster_labels)
```

```python
print("\nSum of Squared Errors (SSE):")
print(sse)
```

```python
plt.scatter(data_values[:, 0], data_values[:, 1], c=cluster_labels, cmap='viridis', marker='o')
```

```python
plt.scatter(final_centroids[:, 0], final_centroids[:, 1], s=200, c='red', marker='X', label='Centroids')
plt.title(f'k-Means Clustering (k={k})')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show
```
